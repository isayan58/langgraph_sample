```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel, RunnableLambda, RunnableBranch, RunnablePassthrough
import os
from dotenv import load_dotenv

load_dotenv()

llm_openai= init_chat_model(model=os.environ.get("LLM_MODEL"), temperature = 0.7)
# generation = input("Enter your generation (e.g., GenZ, Millennial, GenX): ")
```

```python
from typing import TypedDict, List

class graph_schema(TypedDict):
    name : str
    message : str
```

# Node Function

```python
def welcome(state: graph_schema)->graph_schema:
    curr_name = state['name']
    curr_message = state['message']

    response = llm_openai.invoke(f"My name is {curr_name} and I want to say {curr_message}.").content

    state['message'] = response

    return state
```

```python
# State graph
from langgraph.graph import StateGraph, START, END

graph = StateGraph(graph_schema)

graph.add_node("welcome", welcome)
graph.add_edge(START,"welcome")
graph.add_edge("welcome", END)
```

```python
# Compile the graph
from Ipython.display import Image, display

first_graph = graph.compile()

print(first_graph.get_graph().draw_mermaid_png())
```
